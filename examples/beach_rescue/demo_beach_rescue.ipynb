```python
from fmdtools.sim import propagate
from fmdtools.define.architecture.function import FunctionArchitectureGraph
from fmdtools_examples.beach_rescue.main_model import BeachEnvironment, Drone, Hazard
from IPython.display import HTML
import inspect
from matplotlib.colors import ListedColormap
import matplotlib.patches as mpatches



# Initialize objects safely
env = BeachEnvironment("env")
hazard = Hazard()
hazard.drowning()
drone = Drone()

env.c.update_objects(hazard.s.location, drone.s.location)



# Set visualization parameters
cmap = ListedColormap(["skyblue", "lemonchiffon"])
collections = { "start": {"color": "lime"}, "hazard": {"color": "red"}, "drone": {"color": "black"}}

# Plot map background layout configuration
fig, ax = env.c.show({"terrain": {"cmap": cmap}}, collections=collections)
fig.axes[-1].remove()  # Strip active colorbar axis wrapper

ocean_patch = mpatches.Patch(color="skyblue", label="ocean")
sand_patch = mpatches.Patch(color="lemonchiffon", label="sand")
start_patch = mpatches.Patch(color="lime", label="start")
hazard_patch = mpatches.Patch(color="red", label="hazard")
drone_patch = mpatches.Patch(color="black", label="drone")
ax.legend(handles=[ocean_patch, sand_patch, start_patch, hazard_patch, drone_patch], loc='upper right', bbox_to_anchor=(1.3, 1))
```

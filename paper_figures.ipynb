```python
from fireenvironment import show_properties, sim_properties
from fmdtools.analyze.common import consolidate_legend
from fmdtools.define.object.coords import MetricCoords
from wildfiresim import WildfireSim, WildFireSimParameter, def_p, create_scen_sample
import numpy as np
```

```python
from fmdtools.sim.sample import ParameterDomain
from fmdtools.sim.search import ParameterSimProblem


def create_problem(**samp_kwargs):
    light_mdl = WildfireSim(p=def_p, track=None)
    pd = ParameterDomain(WildFireSimParameter.from_base_loc)
    pd.add_variable('x', var_lim = (0, 45))
    pd.add_variable('y', var_lim = (0, 45))
    pd.add_constant('num_strikes', 4)

    ps = create_scen_sample(**samp_kwargs)
    psp = ParameterSimProblem(light_mdl, pd, 'parameter_sample', ps, keep_ec=True)
    psp.add_result_objective('perc_burned', 'fxns.firepropagation.s.perc_burned', metric=np.mean)
    return psp
```

```python
traj_plot_kwar = dict(time="iter", time_ticks=1,  t_pretext="", time_groups=['nominal'])
```

Fire Propagation Map (with details)

```python
def show_scen_strikes(fig=None, ax=None, color='blue', alpha=0.25, with_map=True, num_strikes=4,
                      **samp_kwargs):
    if with_map:
        mdl = WildfireSim(p={**def_p, 'num_strikes': num_strikes})
        fig, ax = mdl.flows['fireenvironment'].c.show(properties=show_properties,
                                                      fig=fig, ax=ax)
    ps = create_scen_sample(**samp_kwargs)
    props = {'strike': {"color": color, "alpha": alpha}}
    for scen in ps.scenarios():
        mdl = WildfireSim(p={**def_p, 'num_strikes': num_strikes}, r=scen.r, track=None)
        fig, ax = mdl.flows['fireenvironment'].c.show(properties=props, fig=fig, ax=ax)
    return fig, ax
```

```python
# average over all seeds:
fig, ax = show_scen_strikes(seed=10, color="violet", alpha=0.65, replicates=20)
ax.set_title("Map Fuels and Strike Locations")
ax.set_xlabel("x (km)")
ax.set_ylabel("y (km)")
fig.savefig("map_and_strikes.pdf",  bbox_inches='tight')
```

# Optimization 

```python
from scipy.optimize import minimize
from scipy.optimize import fmin_powell

def show_opt(ax, res, marker="*", color="red", text=True, linewidth=2.0, textcolor="red", fontsize=12, digits=3, **kwargs):
    ax.scatter(*res.x, marker=marker, color=color, linewidth=linewidth, zorder=2)
    if text:
        ax.annotate(str(round(res.fun, digits)), res.x, color=textcolor, fontsize=fontsize, **kwargs)
```

### Experiment 0: BFGS

BFGS Struggled at 10 replicates. At 20 replicates...

```python
import autograd.numpy as np
from autograd import grad, jacobian
from scipy.optimize._linesearch import line_search_armijo

def bfgs(x0, fun, grad, grad_threshold=1e-30, max_iter=100, n_linesearch=10):
    x = np.copy(x0)
    path = [np.copy(x0)]

    n_dims = x.shape[0]
    B = np.eye(n_dims)

    old_f = None

    g = grad(x)
    for _ in range(max_iter):
        p = np.linalg.solve(B, -g)

        alpha, n_feval, old_f = line_search_armijo(fun, x, p, g, old_f)
        if alpha is None:  # found no solution that satisfies the conditions
            alpha = 1.0

        s = alpha * p
        x += s

        prev_g = g
        g = grad(x)
        if np.linalg.norm(g) <= grad_threshold:
            break

        if prev_g is not None:
            y = g - prev_g
            B += (np.outer(y, y) / np.dot(y, s) -
                  B.dot(s[:, np.newaxis]).dot(s[np.newaxis, :]).dot(B) / s.dot(B).dot(s))

        path.append(np.copy(x))

    return x, path
```

```python
psp_bfgs = create_problem(replicates=20)
```

```python
approx_grad = grad(psp_bfgs.perc_burned)
```

### Experiment 1: Nelder-Mead

(Basically, it doesn't seem to work because of the weirdness of the objective.)

```python
initial_simplex = [[10.0, 10.0],[10.0, 40.0],[40.0, 40.0]]
psp_nm = create_problem(replicates=5)

nm_res = minimize(psp_nm.perc_burned, [10.0, 10.0], method="Nelder-Mead",
                    bounds=((-1.0, 47.), (-1.0, 47.)),
                    options=dict(initial_simplex=initial_simplex,
                            maxiter=3,
                            disp=True))
    
nm_res
```

### Experiment 3: DIRECT

This worked pretty well at n=10 points and I expect it to work here also.
